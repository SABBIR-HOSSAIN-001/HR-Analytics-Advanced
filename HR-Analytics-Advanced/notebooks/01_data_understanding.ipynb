```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", "{:.2f}".format)

file_path = "../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
satisfaction_attrition = (
    pd.crosstab(
        df["JobSatisfaction"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

satisfaction_attrition
```

```python
environment_attrition = (
    pd.crosstab(
        df["EnvironmentSatisfaction"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

environment_attrition
```

```python
worklife_attrition = (
    pd.crosstab(
        df["WorkLifeBalance"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

worklife_attrition
```

```python
relationship_attrition = (
    pd.crosstab(
        df["RelationshipSatisfaction"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

relationship_attrition
```

```python
df["CareerStage"] = pd.cut(
    df["YearsAtCompany"],
    bins=[-1, 2, 5, 10, float("inf")],
    labels=[
        "Early Career",
        "Developing",
        "Established",
        "Long Tenure"
    ]
)

df["CareerStage"].value_counts()
```

```python
careerstage_attrition = (
    pd.crosstab(
        df["CareerStage"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

careerstage_attrition
```

```python
df["MonthlyIncome"].describe()
```

```python
df["IncomeBand"] = pd.cut(
    df["MonthlyIncome"],
    bins=[0, 3000, 6000, 10000, float("inf")],
    labels=[
        "Low Income",
        "Lower-Mid Income",
        "Upper-Mid Income",
        "High Income"
    ]
)

df["IncomeBand"].value_counts().sort_index()
```

```python
income_attrition = (
    pd.crosstab(
        df["IncomeBand"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

income_attrition
```

```python
income_summary = (
    df.groupby("IncomeBand", observed=True)
      .agg(
          Total_Employees=("EmployeeNumber", "count"),
          Attrition_Employees=("Attrition", lambda x: (x == "Yes").sum())
      )
)

income_summary["Attrition_Rate_%"] = (
    income_summary["Attrition_Employees"]
    / income_summary["Total_Employees"]
    * 100
)

income_summary
```

```python
joblevel_income = (
    df.groupby("JobLevel")
      .agg(
          Employees=("EmployeeNumber", "count"),
          Average_Income=("MonthlyIncome", "mean"),
          Median_Income=("MonthlyIncome", "median")
      )
      .round(2)
)

joblevel_income
```

```python
joblevel_attrition = (
    df.groupby("JobLevel")
      .agg(
          Total_Employees=("EmployeeNumber", "count"),
          Attrition_Employees=("Attrition", lambda x: (x == "Yes").sum())
      )
)

joblevel_attrition["Attrition_Rate_%"] = (
    joblevel_attrition["Attrition_Employees"]
    / joblevel_attrition["Total_Employees"]
    * 100
)

joblevel_attrition.round(2)
```

```python
df["TenureBand"] = pd.cut(
    df["YearsAtCompany"],
    bins=[-1, 2, 5, 10, float("inf")],
    labels=[
        "0-2 Years",
        "3-5 Years",
        "6-10 Years",
        "10+ Years"
    ]
)

df["TenureBand"].value_counts().sort_index()
```

```python
tenure_attrition = (
    pd.crosstab(
        df["TenureBand"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

tenure_attrition.round(2)
```

```python
role_tenure_summary = (
    df.groupby("TenureBand", observed=True)
      .agg(
          Employees=("EmployeeNumber", "count"),
          Avg_YearsAtCompany=("YearsAtCompany", "mean"),
          Avg_YearsInCurrentRole=("YearsInCurrentRole", "mean"),
          Avg_YearsWithManager=("YearsWithCurrManager", "mean")
      )
      .round(2)
)

role_tenure_summary
```

```python
df["RoleProgressionGap"] = (
    df["YearsAtCompany"] - df["YearsInCurrentRole"]
)

df[
    [
        "YearsAtCompany",
        "YearsInCurrentRole",
        "RoleProgressionGap",
        "Attrition"
    ]
].head(10)
```

```python
df["RoleProgressionBand"] = pd.cut(
    df["RoleProgressionGap"],
    bins=[-1, 2, 5, 10, float("inf")],
    labels=["0–2 Years", "3–5 Years", "6–10 Years", "10+ Years"]
)

df["RoleProgressionBand"].value_counts().sort_index()
```

```python
progression_attrition = (
    df.groupby("RoleProgressionBand", observed=True)
      .agg(
          Employees=("EmployeeNumber", "count"),
          Attrition_Count=("Attrition", lambda x: (x == "Yes").sum())
      )
      .reset_index()
)

progression_attrition["Attrition_Rate"] = (
    progression_attrition["Attrition_Count"]
    / progression_attrition["Employees"]
    * 100
).round(2)

progression_attrition
```

```python
promotion_summary = (
    df.groupby("YearsSinceLastPromotion")
      .agg(
          Employees=("EmployeeNumber", "count"),
          Attrition_Count=("Attrition", lambda x: (x == "Yes").sum())
      )
      .reset_index()
)

promotion_summary["Attrition_Rate"] = (
    promotion_summary["Attrition_Count"]
    / promotion_summary["Employees"]
    * 100
).round(2)

promotion_summary
```

```python
df["PromotionDelayBand"] = pd.cut(
    df["YearsSinceLastPromotion"],
    bins=[-1, 2, 5, 10, float("inf")],
    labels=["0–2 Years", "3–5 Years", "6–10 Years", "10+ Years"]
)

df["PromotionDelayBand"].value_counts().sort_index()
```

```python
promotion_delay_attrition = (
    df.groupby("PromotionDelayBand", observed=True)
      .agg(
          Employees=("EmployeeNumber", "count"),
          Attrition_Count=("Attrition", lambda x: (x == "Yes").sum())
      )
      .reset_index()
)

promotion_delay_attrition["Attrition_Rate"] = (
    promotion_delay_attrition["Attrition_Count"]
    / promotion_delay_attrition["Employees"]
    * 100
).round(2)

promotion_delay_attrition
```

```python
career_progression_summary = (
    df.groupby("PromotionDelayBand", observed=True)
      .agg(
          Employees=("EmployeeNumber", "count"),
          Avg_YearsAtCompany=("YearsAtCompany", "mean"),
          Avg_YearsInCurrentRole=("YearsInCurrentRole", "mean"),
          Avg_YearsSincePromotion=("YearsSinceLastPromotion", "mean"),
          Attrition_Rate=("Attrition", lambda x: (x == "Yes").mean() * 100)
      )
      .round(2)
)

career_progression_summary
```

```python
df["Overtime_Risk"] = (df["OverTime"] == "Yes").astype(int)

df["Low_JobSatisfaction"] = (df["JobSatisfaction"] == 1).astype(int)

df["Low_WorkLifeBalance"] = (df["WorkLifeBalance"] == 1).astype(int)

df["Low_JobInvolvement"] = (df["JobInvolvement"] == 1).astype(int)
```

```python
df["Employee_Risk_Score"] = (
    df["Overtime_Risk"]
    + df["Low_JobSatisfaction"]
    + df["Low_WorkLifeBalance"]
    + df["Low_JobInvolvement"]
)
```

```python
risk_summary = (
    df.groupby("Employee_Risk_Score")
      .agg(
          Employees=("EmployeeNumber", "count"),
          Attrition_Count=("Attrition", lambda x: (x == "Yes").sum())
      )
      .reset_index()
)

risk_summary["Attrition_Rate"] = (
    risk_summary["Attrition_Count"]
    / risk_summary["Employees"]
    * 100
).round(2)

risk_summary
```

```python
df["Risk_Segment"] = pd.cut(
    df["Employee_Risk_Score"],
    bins=[-1, 0, 1, 2, 4],
    labels=["Low Risk", "Moderate Risk", "High Risk", "Very High Risk"]
)

df["Risk_Segment"].value_counts()
```

```python
risk_segment_summary = (
    df.groupby("Risk_Segment", observed=True)
      .agg(
          Employees=("EmployeeNumber", "count"),
          Attrition_Count=("Attrition", lambda x: (x == "Yes").sum())
      )
      .reset_index()
)

risk_segment_summary["Attrition_Rate"] = (
    risk_segment_summary["Attrition_Count"]
    / risk_segment_summary["Employees"]
    * 100
).round(2)

risk_segment_summary
```

```python
df["Attrition_Binary"] = (
    df["Attrition"] == "Yes"
).astype(int)
```
